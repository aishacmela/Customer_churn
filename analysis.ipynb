```python
numeric_cols = ['Subscription Length (Months)','Customer Satisfaction Score (1-10)',
                'Daily Watch Time (Hours)', 'Engagement Rate (1-10)',
                 'Support Queries Logged', 'Age', 'Monthly Income ($)',
                 'Promotional Offers Used', 'Number of Profiles Created', 'Churn Numeric']

for col in numeric_cols:
    df[col].hist(bins=20)
    plt.title(col)
    plt.show()
```

Bivariate Analysisis
start relating features to the targert, churn status


```python
for col in numeric_cols:
    df.boxplot(column=col, by='Churn Status')
    plt.title(col)
    plt.suptitle('')
    plt.show()
```

```python
pd.crosstab(df['Subscription Plan'], df['Churn Status'], normalize='index')
```

```python
#Convert Churn status to numeric 

df['Churn Numeric'] = df['Churn Status'].map({'Yes': 1, 'No': 0})
print(df['Churn Numeric'].value_counts())
```

```python
corr =df[numeric_cols].corr()
print(corr)
```

```python
plt.figure(figsize=(12,5))
sns.heatmap(corr, 
            annot=True,
            cmap='coolwarm',
            vmin=-1,
            vmax=1,
            center=0)
plt.show
```

Given weak individual predictors across the board, this is actually a useful signal about what kind of model will work here: tree-based models (Random Forest, Gradient Boosting) tend to do better than linear models (like Logistic Regression) in cases like this, since they can capture non-linear thresholds and feature interactions that a correlation matrix — and a plain logistic regression
